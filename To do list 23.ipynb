```python
todo_list=[]
completed=[]

def addd():
    z= input ("enter your new to do list here:")
    todo_list.append (z)

def edit():
    showlist()
    old= input("Enter the task you want to edit:")
    if old in todo_list:
        new= input("Enter the new name for the task:")
        todo_list.remove(old)
        todo_list.append(new)
        print("Task updated")
    else:
        print("Task not found")
    
def comp ():
    showlist ()
    h= input ("enter your completed task:")
    todo_list.remove (h)
    completed.append(h)

def rem():
    showlist ()
    y=input ("enter the task you want to remove")
    todo_list.remove (y)

def showlist ():
    print (todo_list) 


def showcomp():
    print (completed)


def todo():
    print("====TO DO LIST====")
    showlist ()
    print("===COMPLETED===")
    showcomp()
    print("===MAIN MENU===")
    print("1 to add new to do list")
    print("2 to remove a task from the list")
    print("3 to mark as completed")
    print("4 to edit your list")
    print("5 to exit the app")
    c=int(input("enter a number from 1 to 5:"))
    if c==1:
        addd()
    elif c==2:
        rem()
    elif c==3:
        comp()
    elif c==4:
        edit()
    elif c==5:
        print("exiting the app")
        return "EXIT"
    else:
        print ("Invalid input")
        



while True:
    if todo() == "EXIT":
        break
    todo()
```
